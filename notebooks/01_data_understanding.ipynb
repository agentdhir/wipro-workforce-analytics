```python
from pathlib import Path
import pandas as pd
import numpy as np

project_root = Path.cwd().parent
processed_path = project_root / "data" / "processed"

combined_file_path = processed_path / "combined_raw_staging.csv"
no_duplicate_file_path = processed_path / "combined_no_exact_duplicates.csv"

print("Project root:", project_root)
print("Processed path:", processed_path)
print("Combined raw staging exists:", combined_file_path.exists())
print("No exact duplicate file exists:", no_duplicate_file_path.exists())

if combined_file_path.exists():
    print("Combined file size MB:", combined_file_path.stat().st_size / (1024 * 1024))

if no_duplicate_file_path.exists():
    print("No duplicate file size MB:", no_duplicate_file_path.stat().st_size / (1024 * 1024))
```

```python
# Load combined raw staging file
df = pd.read_csv(combined_file_path, low_memory=False)

print("Combined raw staging loaded")
print("Shape before duplicate removal:", df.shape)

# Remove exact duplicate rows
df_no_exact_duplicates = df.drop_duplicates().copy()

print("Shape after exact duplicate removal:", df_no_exact_duplicates.shape)
print("Exact duplicate rows removed:", df.shape[0] - df_no_exact_duplicates.shape[0])

# Save no-exact-duplicate file
df_no_exact_duplicates.to_csv(no_duplicate_file_path, index=False)

print("No exact duplicate file saved at:")
print(no_duplicate_file_path)
```
